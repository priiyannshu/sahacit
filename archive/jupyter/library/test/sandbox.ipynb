### 🧭 Why the Gradient Points in the Direction of Steepest Ascent

To understand why the gradient vector $\nabla f(\mathbf{x})$ points in the direction where a multivariable function increases most rapidly, we look at the **Directional Derivative** and the geometry of the **Dot Product**.

---

### 1. The Directional Derivative {\mathbf{u}} f(\mathbf{x})$

Suppose you are standing on a surface described by a scalar field  = f(x, y)$ at point $\mathbf{x} = (x, y)$.
If you step in any direction given by a **unit vector** $\mathbf{u}$ (where $\|\mathbf{u}\| = 1$), the instantaneous rate of change of $ along $\mathbf{u}$ is defined as:

$$D_{\mathbf{u}} f(\mathbf{x}) = \lim_{h \to 0} \frac{f(\mathbf{x} + h\mathbf{u}) - f(\mathbf{x})}{h}$$

By the multivariable chain rule (or first-order Taylor approximation):
$$f(\mathbf{x} + h\mathbf{u}) \approx f(\mathbf{x}) + \nabla f(\mathbf{x}) \cdot (h\mathbf{u})$$
$$\implies D_{\mathbf{u}} f(\mathbf{x}) = \nabla f(\mathbf{x}) \cdot \mathbf{u}$$

The directional derivative is simply the **dot product** of the gradient vector $\nabla f(\mathbf{x})$ and your direction vector $\mathbf{u}$.

---

### 2. The Geometric Dot Product

Recall the geometric formula for the dot product between two vectors $\mathbf{a}$ and $\mathbf{b}$:
$$\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \,\|\mathbf{b}\| \cos(\theta)$$

where $\theta$ is the angle between $\mathbf{a}$ and $\mathbf{b}$. Applying this to the directional derivative:

$$D_{\mathbf{u}} f(\mathbf{x}) = \nabla f(\mathbf{x}) \cdot \mathbf{u} = \|\nabla f(\mathbf{x})\| \,\|\mathbf{u}\| \cos(\theta)$$

Since $\mathbf{u}$ is a unit vector ($\|\mathbf{u}\| = 1$):

$$D_{\mathbf{u}} f(\mathbf{x}) = \|\nabla f(\mathbf{x})\| \cos(\theta)$$

At any fixed point $\mathbf{x}$, the gradient $\nabla f(\mathbf{x})$ is fixed, so its magnitude $\|\nabla f(\mathbf{x})\|$ is a constant scalar. Therefore, the rate of change depends **entirely on $\cos(\theta)*!

---

### 3. Finding the Extremes

Because the cosine function satisfies hBc1 \le \cos(\theta) \le 1$:

1. **Maximum Rate of Increase (Steepest Ascent)**:
   - $\cos(\theta)$ achieves its maximum value of $+1$ when **$\theta = 0^\circ*.
   - $\theta = 0$ means the direction $\mathbf{u}$ is **parallel and aligned** with $\nabla f(\mathbf{x})$:
     $$\mathbf{u}^*_{\text{ascent}} = \frac{\nabla f(\mathbf{x})}{\|\nabla f(\mathbf{x})\|}$$
   - The maximum rate of increase equals the magnitude of the gradient: $\|\nabla f(\mathbf{x})\|$.

2. **Maximum Rate of Decrease (Steepest Descent)**:
   - $\cos(\theta) = -1$ when **$\theta = 180^\circ$ ($\pi$ rad)**.
   - $\mathbf{u}$ points directly **opposite** to $\nabla f$:
     $$\mathbf{u}^*_{\text{descent}} = -\frac{\nabla f(\mathbf{x})}{\|\nabla f(\mathbf{x})\|}$$
   - This is why **Gradient Descent** updates weights as $\mathbf{w} \leftarrow \mathbf{w} - \eta \nabla L$.

3. **No Change (Contour / Level Curves)**:
   - When $\theta = 90^\circ$ ($\cos(\theta) = 0$), {\mathbf{u}} f(\mathbf{x}) = 0$.
   - This proves the gradient vector is always **perpendicular (orthogonal)** to the level curves (contours) of constant value.

### 💻 Numerical Verification:
Run the code cell below to sweep through all angles and verify that the directional derivative is maximized precisely when aligned with $\nabla f$.

```python
# Numerical Verification: Directional Derivative vs Angle theta
import math

# Define paraboloid f(x, y) = 3x^2 + 2y^2
# Analytical gradient: grad f = [df/dx, df/dy] = [6x, 4y]
x, y = 1.0, 1.0
grad_x = 6.0 * x  # 6.0
grad_y = 4.0 * y  # 4.0
grad_mag = math.hypot(grad_x, grad_y)
grad_angle_deg = math.degrees(math.atan2(grad_y, grad_x))

print(f"Point (x, y): ({x}, {y})")
print(f"Gradient vector: [{grad_x:.2f}, {grad_y:.2f}]")
print(f"Gradient magnitude ||grad f||: {grad_mag:.4f}")
print(f"Gradient angle: {grad_angle_deg:.2f}°
")

# Test directions around 360 degrees
print(f"{'Angle':>6} | {'Unit Vector (ux, uy)':>22} | {'cos(theta)':>10} | {'D_u f (Rate)':>12}")
print("-" * 58)

best_deg = None
max_rate = -float('inf')

for deg in range(0, 361, 15):
    rad = math.radians(deg)
    ux, uy = math.cos(rad), math.sin(rad)
    
    # Directional derivative = dot product (grad . u)
    rate = grad_x * ux + grad_y * uy
    cos_theta = rate / grad_mag
    
    if rate > max_rate:
        max_rate = rate
        best_deg = deg
        
    # Print key angles
    if deg % 45 == 0 or abs(deg - round(grad_angle_deg)) < 8:
        tag = " <-- Peak (Steepest Ascent)!" if abs(rate - grad_mag) < 0.05 else ""
        print(f"{deg:5d}° | ({ux:8.4f}, {uy:8.4f}) | {cos_theta:10.4f} | {rate:12.4f}{tag}")

print("-" * 58)
print(f"✅ Max rate of change: {max_rate:.4f} (at {best_deg}°)")
print(f"✅ Matches ||grad f|| ({grad_mag:.4f}) at angle {grad_angle_deg:.2f}°!")
```
